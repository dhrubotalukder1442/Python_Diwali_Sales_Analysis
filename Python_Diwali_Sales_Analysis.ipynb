```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt # visualizing data
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv("E:\\Diwali Sales Data.csv",encoding= 'unicode_escape')
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
```

```python
pd.isnull(df).sum()
```

```python
# drop null values
df.dropna(inplace=True)
```

```python
# change data type
df['Amount'] = df['Amount'].astype('int')
```

```python
df['Amount'].dtypes
```

```python
df.columns
```

```python
#rename column
df.rename(columns= {'Marital_Status':'Shaadi'})
```

```python
# describe() method returns description of the data in the DataFrame (i.e. count, mean, std, etc)
df.describe()
```

```python
# use describe() for specific columns
df[['Age', 'Orders', 'Amount']].describe()
```

```python
# plotting a bar chart for Gender and it's count

gender_counts = df['Gender'].value_counts()
gender_counts.plot(kind='bar', color=['blue', 'pink'])
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Gender Count')
plt.show()
```

```python
# plotting a bar chart for gender vs total amount

# Grouping the data by gender and calculating the total amount for each gender
gender_total_amount = df.groupby('Gender')['Amount'].sum()

# Plotting the bar chart
gender_total_amount.plot(kind='bar', color=['pink', 'green'])
plt.xlabel('Gender')
plt.ylabel('Total Amount')
plt.title('Total Amount by Gender')
plt.show()
```

```python

ax = sns.countplot(data=df, x='Age Group', hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)

# Displaying the plot
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.title('Count of Age Groups by Gender')
plt.show()
```

```python
# Total Amount vs Age Group
age_group_total_amount = df.groupby('Age Group')['Amount'].sum()
age_group_total_amount.plot(kind='bar', color='skyblue')
plt.xlabel('Age Group')
plt.ylabel('Total Amount')
plt.title('Total Amount by Age Group')
plt.show()
```

```python
# total number of orders from top 10 states

sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State',y= 'Orders')
```

```python
# total amount/sales from top 10 states

state_order_counts = df['State'].value_counts().nlargest(10)

# Plotting the bar chart
state_order_counts.plot(kind='bar')
plt.xlabel('State')
plt.ylabel('Number of Orders')
plt.title('Top 10 States by Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
ax = sns.countplot(data = df, x = 'Marital_Status')

sns.set(rc={'figure.figsize':(7,5)})
for bars in ax.containers:
    ax.bar_label(bars)
    # Displaying the plot
plt.xlabel('Marital Status')
plt.ylabel('Count')
plt.title('Marital Status Count')
plt.show()
```

```python
sales_state = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(data = sales_state, x = 'Marital_Status',y= 'Amount', hue='Gender')
```

```python
# Setting the figure size
sns.set(rc={'figure.figsize':(15,13)})

# Creating the countplot
ax = sns.countplot(data=df, x='Occupation')

# Adding labels to the bars
for bars in ax.containers:
    ax.bar_label(bars)

# Displaying the plot
plt.xlabel('Occupation')
plt.ylabel('Count')
plt.title('Occupation Count')
plt.show()
```

```python
sales_state = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

# Setting the figure size
sns.set(rc={'figure.figsize':(20,5)})

# Creating the bar plot
sns.barplot(data=sales_state, x='Occupation', y='Amount')

# Displaying the plot
plt.xlabel('Occupation')
plt.ylabel('Total Amount')
plt.title('Total Amount by Occupation')
plt.show()
```

```python
# Setting the figure size
sns.set(rc={'figure.figsize':(20,5)})

# Creating the countplot
ax = sns.countplot(data=df, x='Product_Category')

# Adding labels to the bars
for bars in ax.containers:
    ax.bar_label(bars)

# Displaying the plot
plt.xlabel('Product Category')
plt.ylabel('Count')
plt.title('Product Category Count')
plt.show()
```

```python
# Grouping the data by product category and calculating the total amount for each category
sales_state = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

# Setting the figure size
sns.set(rc={'figure.figsize':(20,5)})

# Creating the bar plot
sns.barplot(data=sales_state, x='Product_Category', y='Amount')

# Displaying the plot
plt.xlabel('Product Category')
plt.ylabel('Total Amount')
plt.title('Total Amount by Top 10 Product Categories')
plt.show()
```

```python
sales_state = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

# Setting the figure size
sns.set(rc={'figure.figsize':(20,5)})

# Creating the bar plot
sns.barplot(data=sales_state, x='Product_ID', y='Orders')

# Displaying the plot
plt.xlabel('Product ID')
plt.ylabel('Total Orders')
plt.title('Total Orders by Top 10 Product IDs')
plt.show()
```
